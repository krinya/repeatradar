```python
import pandas as pd
```

# Load Data

```python
ecommerce_data = pd.read_csv('ecommerce_data_2.csv')
```

```python
ecommerce_data.head()
```

```python
# shape of the data
print("Shape of the data:", ecommerce_data.shape)
```

```python
# pring unique Customer_Id count
print("Unique Customer_Id count:", ecommerce_data['Customer_Id'].nunique())
```

# Fix and create some columns

```python
ecommerce_data['OderedDateTime'] = ecommerce_data['Order_Date'] + ' ' + ecommerce_data['Time']
# convert to datetime
ecommerce_data['OderedDateTime'] = pd.to_datetime(ecommerce_data['OderedDateTime'], format='%Y-%m-%d %H:%M:%S')
```

```python
# count NaN values in each column
nan_counts = ecommerce_data.isna().sum()
# print NaN counts
print(nan_counts)
```

```python
# just to make sure we have no NaN values in OderedDateTime and CustomerID
ecommerce_data = ecommerce_data.dropna(axis = 0, subset = ['OderedDateTime'])
ecommerce_data = ecommerce_data.dropna(axis = 0, subset = ['Customer_Id'])
```

```python
ecommerce_data.drop_duplicates(inplace = True)
```
